# ChipWhisperer Nano – Windows
Native Windows-Python-Installation; keine venv, kein WSL. Die Verbindung konfiguriert das Nano mit default_setup(), flasht aber keine Firmware. Vor Ausfuehrung andere ChipWhisperer-Verbindungen schliessen.

```python
import chipwhisperer as cw
print(cw.__version__)
print(cw.list_devices())
```

```python
scope = cw.scope(scope_type=cw.scopes.CWNano)
try:
    scope.default_setup()
    target = cw.target(scope)
except Exception:
    scope.dis()
    raise
print(scope)
```
